```python
x=[35,66,78,97,46,98,41,32,69,95]
print(x)

```

```python
#find maximum
max=x[0]
for i in range(1,len(x),1):
    if x[i] > max:
        max=x[i]
        id = i
        print(max,i)
    else:
        continue
print(f"the maximum is {max} at index {id}")       
```

```python
max = x[0]
for j in x:
    if j>max:
        max=j
print(f"the maximum is {max}")
```

```python
#find minimum
min=x[0]
for i in range(1,len(x),1):
    if x[i] < min:
        min=x[i]
        id = i
        print(min,i)
    else:
        continue
print(f"the minimum is {min} at index {id}")        
        
```
